---

## Inserção de dados

```python
db = client["AulaDemo"]
colecao = db["Estudantes"]

dados = [
    {"nome": "Thiago", "idade": 21, "curso": "Software"},
    {"nome": "Ana", "idade": 22, "curso": "ADS"},
    {"nome": "Carlos", "idade": 25, "curso": "CC"}
]

colecao.insert_many(dados)
```

---

## Consultas

```python
list(colecao.find())
list(colecao.find({"curso": "ADS"}))
list(colecao.find().sort("nome", 1))
list(colecao.find().limit(2))
```

---

## Atualização e remoção

```python
colecao.update_one({"nome": "Ana"}, {"$set": {"idade": 23}})
colecao.delete_one({"nome": "Carlos"})
```

---

## Criação de índice

```python
colecao.create_index("nome")
```

---

## Aggregation + DataFrame

```python
import pandas as pd

pipeline = [
    {"$group": {"_id": "$curso", "total": {"$sum": 1}}}
]

df = pd.DataFrame(list(colecao.aggregate(pipeline)))
print(df)
```

---

## Coleção Vendas e gráfico

```python
import matplotlib.pyplot as plt

vendas = db["Vendas"]

vendas.insert_many([
    {"produto": "Notebook", "categoria": "Eletrônicos", "valor": 3000},
    {"produto": "Mouse", "categoria": "Eletrônicos", "valor": 150},
    {"produto": "Cadeira", "categoria": "Móveis", "valor": 800},
    {"produto": "Mesa", "categoria": "Móveis", "valor": 1200}
])

pipeline = [
    {"$group": {"_id": "$categoria", "total": {"$sum": "$valor"}}}
]

df = pd.DataFrame(list(vendas.aggregate(pipeline)))

df.plot(x="_id", y="total", kind="bar")
plt.title("Receita por Categoria")
plt.xlabel("Categoria")
plt.ylabel("Total")
plt.show()
```

# 4. Questão 3 — Docker Compose

## Arquivo docker-compose.yml

```yaml
version: '3.8'

services:

  mongodb:
    image: mongo
    ports:
      - "27017:27017"
    volumes:
      - mongo_data:/data/db
    networks:
      - mybridge

  redis:
    image: redis
    ports:
      - "6379:6379"
    networks:
      - mybridge

  neo4j:
    image: neo4j
    ports:
      - "7474:7474"
      - "7687:7687"
    environment:
      - NEO4J_AUTH=neo4j/test
    volumes:
      - neo4j_data:/data
    networks:
      - mybridge

  jupyter:
    image: jupyter/base-notebook
    ports:
      - "8888:8888"
    networks:
      - mybridge

volumes:
  mongo_data:
  neo4j_data:

networks:
  mybridge:
```

---

## Execução

```bash
docker-compose up -d
```

---

## Verificação

```bash
docker ps
```

Foram executados com sucesso os seguintes containers:

* MongoDB
* Redis
* Neo4j
* Jupyter

# 5. Questão 4 — Análise Exploratória de Dados

## Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt

dados = {
    "regiao": ["Centro-Oeste", "Sudeste", "Sudeste", "Centro-Oeste", "Sudeste"],
    "estado": ["DF", "SP", "RJ", "DF", "SP"],
    "doc_fem": [100, 200, 150, 120, 180],
    "doc_masc": [120, 220, 160, 130, 200]
}

df = pd.DataFrame(dados)
```

---

## Instituições por região

```python
df["regiao"].value_counts().plot(kind="bar")
plt.title("Instituições por Região")
plt.show()
```

---

## Docentes por gênero

```python
plt.bar(
    ["Feminino", "Masculino"],
    [df["doc_fem"].sum(), df["doc_masc"].sum()]
)
plt.title("Docentes por Gênero")
plt.show()
```

---

## Instituições por estado

```python
df["estado"].value_counts().plot(kind="bar")
plt.title("Instituições por Estado")
plt.show()
```
