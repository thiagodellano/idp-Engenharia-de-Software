```
import pandas as pd
import matplotlib.pyplot as plt

# Dados simulados (censo)
dados = {
    "regiao": ["Centro-Oeste", "Sudeste", "Sudeste", "Centro-Oeste", "Sudeste"],
    "estado": ["DF", "SP", "RJ", "DF", "SP"],
    "doc_fem": [100, 200, 150, 120, 180],
    "doc_masc": [120, 220, 160, 130, 200]
}

df = pd.DataFrame(dados)

# 1. Instituições por região
df["regiao"].value_counts().plot(kind="bar")
plt.title("Instituições por Região")
plt.show()

# 2. Docentes por gênero
fem = df["doc_fem"].sum()
masc = df["doc_masc"].sum()

plt.bar(["Feminino", "Masculino"], [fem, masc])
plt.title("Docentes por Gênero")
plt.show()

# 3. Instituições por estado
df["estado"].value_counts().plot(kind="bar")
plt.title("Instituições por Estado")
plt.show()
```
